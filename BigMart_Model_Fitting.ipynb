```python
import numpy as np
import pandas as pd
from sklearn.model_selection import cross_val_score
```

```python
df=pd.read_csv('D:\Data Science\Big_Mart_Sales_Prediction\Preprocessed_Data.csv')
df.head()
```

```python
df.columns
```

```python
x=df.drop(['Item_Outlet_Sales','Unnamed: 0','Outlet_Establishment_Year','Item_Fat_Content_Low Fat','Item_Type_Baking Goods','Outlet_Type_Grocery Store'],axis='columns')
y=df['Item_Outlet_Sales']
```

```python
x
```

```python
x.columns
```

```python
y
```

### **Model Fitting**

#### **1. Linear Regression**

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
score_lr=cross_val_score(lr,X=x,y=y,cv=5)
```

```python
print(f"Score for Linear Regression Model : {np.mean(score_lr)}")
```

#### **2. Lasso Regression**

```python
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV
lasso_parameter={
    'alpha':[0.0001,0.0005,0.0009,0.01,0.05,0.09,0.1,0.5,0.9]
}
lasso_reg=GridSearchCV(Lasso(),param_grid=lasso_parameter,cv=5,n_jobs=-1,verbose=1,return_train_score=False)
lasso_reg.fit(x,y)
```

```python
print(f'Best score : {lasso_reg.best_score_}')
print(f'Best Parameter : {lasso_reg.best_params_}')
```

#### **3. Ridge Regression**

```python
from sklearn.linear_model import Ridge
ridge_parameters={
    'alpha':[0.0001,0.0005,0.0009,0.01,0.05,0.09,0.1,0.5,0.9]
}
ridge_reg=GridSearchCV(Ridge(),param_grid=ridge_parameters,cv=5,n_jobs=-1,verbose=1,return_train_score=False)
ridge_reg.fit(x,y)
```

```python
print(f'Best Score : {ridge_reg.best_score_}')
print(f'Best Parameter : {ridge_reg.best_params_}')
```

#### **4. Random Forest Regressor**

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV

rf_parameter={
    'n_estimators':[100,150,200,250],
    'max_depth': [5,6,7,8],
    'min_samples_split': [8,9,10,11],
    'min_samples_leaf': [20,22,24,25]
}
random_forest=RandomizedSearchCV(RandomForestRegressor(),param_distributions=rf_parameter,n_iter=100,verbose=2,return_train_score=False)
random_forest.fit(x,y)
```

```python
print(f'Best Score : {random_forest.best_score_}')
print(f'Best Parameter : {random_forest.best_params_}')
```

#### **XGBoost Regressor**

```python
from xgboost import XGBRegressor
from sklearn.model_selection import RandomizedSearchCV
xgb_params={
    'n_estimators':[12,13,15,16,17,18,19,20,22,24,25],
    'max_depth':[4,5,6,7],
    'booster':['gbtree'],
    'min_child_weight':[60,65,72,75,76,78,80],
    'base_score':[3,4,5,6,7,8],
    'lambda':[2,4,6,8,10,12],
    'alpha':[3,6,8,9,12,14]
}
xgb=RandomizedSearchCV(XGBRegressor(),param_distributions=xgb_params,cv=8,n_iter=100,n_jobs=-1,verbose=2,return_train_score=False)
xgb.fit(X=x,y=y)
```

```python
print(f'Best Score : {xgb.best_score_}')
print(f'Best Parameters : {xgb.best_params_}')
```

From the above model We can clearly see that Random Forest Regressor method gives the highest R2 score 71.33% among all the models.
We will select Random Forest Regressor {n_estimators': 100, 'min_samples_split': 9, 'min_samples_leaf': 25, 'max_depth': 6} for this dataset
